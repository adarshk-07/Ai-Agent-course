```python
# to set up the agent
import dotenv
import os

from openai import  OpenAI
dotenv.load_dotenv()


if not os.environ.get("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY is missing")

print(
    OpenAI().responses.create(
        model = os.environ["OPENAI_DEFAULT_MODEL"],
        input="Say: We are up and running!"

    )
    .output_text
)   


```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
Study_Planner =Agent(
    name="PlanDude",
    instructions ="""
    You are My Mentor
    Make planner which reallistic 
    and Achievable
    Keep it concise and to the point
    """
)
```

```python
prompt = f"""
    Subjects: pandas
    Available hours: 2 hrs
    Goal: to complete revision 

    Make a practical study timetable with breaks and revision.
    """
```

```python
with trace("Study Planner"):
        response = Runner.run_streamed(Study_Planner, prompt)

async for event in response.stream_events():
            if event.type == "raw_response_event" and isinstance( event.data, ResponseTextDeltaEvent):
                    print(event.data.delta, end="", flush=True)
```
